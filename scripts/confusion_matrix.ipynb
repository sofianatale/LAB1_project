```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix
import numpy as np

def plot_confusion_matrix(tp, fn, fp, tn, set_name):
    cm = np.array([[tn, fp],
                   [fn, tp]])

# Etichette quadranti
    group_names = ['True Negative', 'False Positive', 'False Negative', 'True Positive']
    group_counts = [f"{value}" for value in cm.flatten()]
    labels = [f"{name}\n{count}" for name, count in zip(group_names, group_counts)]
    labels = np.asarray(labels).reshape(2, 2)

# PLOT
    plt.figure(figsize=(8,6))
    sns.heatmap(cm, annot=labels, fmt='', cmap='YlGnBu', cbar=False,
                linewidths=1.5, linecolor='black',
                xticklabels=['Negative', 'Positive'],
                yticklabels=['Negative', 'Positive'],
                annot_kws={"fontsize": 12, "weight": "bold"}, square=True)
    plt.xlabel('Predicted Label', fontsize=13, weight='bold')
    plt.ylabel('True Label', fontsize=13, weight='bold')
    plt.title(f'CONFUSION MATRIX – {set_name}', fontsize=15, weight='bold')
    plt.tight_layout()
    plt.savefig(f'confusion_matrix_{set_name.lower().replace(" ", "_")}.png')
    plt.show()

# SET 1
plot_confusion_matrix(tp=180, fn=3, fp=0, tn=283974, set_name="Set 1")

# SET 2
plot_confusion_matrix(tp=182, fn=1, fp=0, tn=283947, set_name="Set 2")
```
